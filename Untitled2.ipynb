```python
import asyncio
asyncio.set_event_loop_policy(asyncio.WindowsProactorEventLoopPolicy())
```

```python
from browser_use import Agent
from browser_use.llm import ChatOllama

llm = ChatOllama(
    model="llama3:8b",
)

agent = Agent(
    task="Open google and search for AI engineering internships",
    llm=llm,
)

await agent.run()
```
